**CREATING FRAUD FEATURES**

```python
import numpy as np

cc["hour"] = (cc["time"] // 3600) % 24
cc["amount_log"] = np.log1p(cc["amount"])
cc.head()
```

```python
aml_trans
```

```python
aml_trans = aml_trans.copy()

aml_trans.columns = [
    "timestamp", "from_bank", "account", "to_bank", "account_1",
    "amount_received", "receiving_currency", "amount_paid",
    "payment_currency", "payment_format", "is_laundering"
]

aml_trans["timestamp"] = pd.to_datetime(aml_trans["timestamp"], errors="coerce")
aml_trans["hour"] = aml_trans["timestamp"].dt.hour
aml_trans["same_currency"] = (aml_trans["receiving_currency"] == aml_trans["payment_currency"]).astype(int)
aml_trans["cross_bank"] = (aml_trans["from_bank"] != aml_trans["to_bank"]).astype(int)
aml_trans["amount_ratio"] = aml_trans["amount_paid"] / aml_trans["amount_received"].replace(0, pd.NA)

aml_trans.head()
```

**AML summary features**




-This gives a basic risk view per account

```python
aml_summary = aml_trans.groupby("account").agg(
    total_paid=("amount_paid", "sum"),
    total_received=("amount_received", "sum"),
    txn_count=("timestamp", "count"),
    avg_paid=("amount_paid", "mean"),
    laundering_count=("is_laundering", "sum")
).reset_index()

aml_summary["laundering_rate"] = aml_summary["laundering_count"] / aml_summary["txn_count"]
aml_summary.head()
```

**Find suspicious accounts**

```python
aml_summary.sort_values("laundering_rate", ascending=False).head(10)
```

### Initial Insights
- Credit card fraud is highly imbalanced.

- Time and amount features may help identify suspicious activity.

### Analysis
- The credit card dataset is highly imbalanced, so fraud is rare and needs careful interpretation.

- The AML dataset now has account-level summaries that help highlight suspicious behaviour.

- The laundering_rate ranking is a useful first-pass risk filter.

- amount, time, and transaction patterns are the main starting features for investigation.


**Rank suspicious accounts**

```python
aml_summary.sort_values("laundering_rate", ascending=False).head(10)
```

**Risk Flag**

```python
aml_summary["risk_level"] = pd.cut(
    aml_summary["laundering_rate"],
    bins=[-0.01, 0.05, 0.20, 1.00],
    labels=["Low", "Medium", "High"]
)

aml_summary.head()
```

### Key Findings
- Credit card fraud is highly imbalanced.
